### Data Variables Explanation

*   **Age**: Age of the employee.
*   **Attrition**: Whether the employee left the company (Yes/No). This is likely the target variable.
*   **BusinessTravel**: Frequency of business travel for the employee (e.g., 'Travel_Rarely', 'Travel_Frequently', 'Non-Travel').
*   **DailyRate**: Daily rate of pay for the employee.
*   **Department**: Department the employee works in (e.g., 'Sales', 'Research & Development', 'Human Resources').
*   **DistanceFromHome**: Distance from home to work (in miles).
*   **Education**: Level of education (e.g., 1 'Below College', 2 'College', 3 'Bachelor', 4 'Master', 5 'Doctor').
*   **EducationField**: Field of education (e.g., 'Life Sciences', 'Medical', 'Marketing', 'Technical Degree', 'Human Resources', 'Other').
*   **EmployeeCount**: Number of employees (always 1 for each row, can be dropped).
*   **EmployeeNumber**: Unique identifier for each employee.
*   **EnvironmentSatisfaction**: Satisfaction with the work environment (1 'Low' to 4 'Very High').
*   **Gender**: Gender of the employee (Female/Male).
*   **HourlyRate**: Hourly rate of pay for the employee.
*   **JobInvolvement**: Level of job involvement (1 'Low' to 4 'Very High').
*   **JobLevel**: Level of job within the company (1 'Entry' to 5 'Executive').
*   **JobRole**: Role of the employee (e.g., 'Sales Executive', 'Research Scientist', 'Laboratory Technician').
*   **JobSatisfaction**: Level of job satisfaction (1 'Low' to 4 'Very High').
*   **MaritalStatus**: Marital status of the employee (e.g., 'Single', 'Married', 'Divorced').
*   **MonthlyIncome**: Monthly income of the employee.
*   **MonthlyRate**: Monthly rate of pay for the employee.
*   **NumCompaniesWorked**: Number of companies the employee has worked for prior to this one.
*   **Over18**: Whether the employee is over 18 years old (always 'Y', can be dropped).
*   **OverTime**: Whether the employee works overtime (Yes/No).
*   **PercentSalaryHike**: Percentage increase in salary last year.
*   **PerformanceRating**: Performance rating of the employee (1 'Low' to 4 'Outstanding').
*   **RelationshipSatisfaction**: Satisfaction with relationships at work (1 'Low' to 4 'Very High').
*   **StandardHours**: Standard hours of work (always 80, can be dropped).
*   **StockOptionLevel**: Level of stock option granted.
*   **TotalWorkingYears**: Total number of years the employee has worked.
*   **TrainingTimesLastYear**: Number of times training was conducted last year.
*   **WorkLifeBalance**: Work-life balance rating (1 'Bad' to 4 'Best').
*   **YearsAtCompany**: Total number of years the employee has been with the current company.
*   **YearsInCurrentRole**: Number of years in the current role.
*   **YearsSinceLastPromotion**: Number of years since the last promotion.
*   **YearsWithCurrManager**: Number of years with the current manager.

# check for outliers

### Outlier Detection using Box Plots

We will create box plots for each numerical feature to visualize their distributions and identify any potential outliers.

```python
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Determine the number of rows and columns for the subplot grid dynamically
num_numerical_cols = len(numerical_cols)
n_cols = 4 # Number of columns in the subplot grid
n_rows = (num_numerical_cols + n_cols - 1) // n_cols # Calculate number of rows needed

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 4, n_rows * 3), constrained_layout=True)
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)
    axes[i].set_ylabel('') # Remove y-label for cleaner plots

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle('Box Plots of Numerical Features for Outlier Detection', y=1.02, fontsize=16) # Add a main title
plt.show()
```

### Interpretation of Numerical Features Box Plots

The box plots for numerical features show the distribution and potential outliers for each variable. Several features, such as `MonthlyIncome`, `NumCompaniesWorked`, `TotalWorkingYears`, `YearsAtCompany`, `YearsInCurrentRole`, `YearsSinceLastPromotion`, and `YearsWithCurrManager`, exhibit a number of outliers, particularly on the higher end. This suggests that there are employees with significantly higher values in these aspects compared to the majority. Other features like `Age`, `DailyRate`, and `HourlyRate` show more symmetrical distributions with fewer extreme outliers.

### Attrition Rate by Categorical Variables

To understand how attrition varies across different categorical features, we will generate count plots for each categorical column, segmented by 'Attrition' (Yes/No). This allows us to visually inspect the relationship between each categorical variable and employee attrition.

```python
# Get all categorical columns except 'Attrition' itself
categorical_cols = df.select_dtypes(include='object').columns.tolist()
categorical_cols.remove('Attrition') # Remove the target variable

# Determine the number of rows and columns for the subplot grid dynamically
num_categorical_cols = len(categorical_cols)
n_cols = 3 # Number of columns in the subplot grid
n_rows = (num_categorical_cols + n_cols - 1) // n_cols # Calculate number of rows needed

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 6, n_rows * 5), constrained_layout=True)
axes = axes.flatten()

for i, col in enumerate(categorical_cols):
    sns.countplot(x=col, hue='Attrition', data=df, ax=axes[i], palette='viridis')
    axes[i].set_title(f'Attrition by {col}')
    axes[i].set_xlabel('') # Remove x-label for cleaner plots
    axes[i].tick_params(axis='x', labelrotation=45) # Rotate x-axis labels

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle('Attrition Distribution Across Categorical Features', y=1.02, fontsize=16)
plt.show()
```

### Interpretation of Attrition by Categorical Variables

From the count plots comparing Attrition across categorical variables, we can observe several interesting patterns:

*   **BusinessTravel:** Employees who travel `Frequently` show a higher proportion of attrition compared to those who travel `Rarely` or `Non-Travel`.
*   **Department:** The `Sales` and `Research & Development` departments have a higher absolute number of employees, but when looking at the proportion, `Human Resources` also shows a noticeable attrition rate.
*   **EducationField:** Employees in `Life Sciences` and `Medical` fields constitute a large portion of the workforce, and while their absolute attrition numbers are higher, other fields like `Technical Degree` or `Marketing` might have relatively high attrition rates when considering their smaller group sizes.
*   **Gender:** The attrition rates appear somewhat similar between `Male` and `Female` employees in proportion to their total numbers.
*   **JobRole:** `Sales Representatives` and `Laboratory Technicians` show a higher propensity for attrition.
*   **MaritalStatus:** `Single` employees appear to have a higher attrition rate compared to `Married` or `Divorced` employees.
*   **OverTime:** Employees who work `Yes` (overtime) show a significantly higher proportion of attrition compared to those who work `No` (no overtime).

### Attrition Rate by Numerical Variables

To analyze the relationship between numerical features and employee attrition, we will create box plots. Each box plot will show the distribution of a numerical variable for employees who stayed (Attrition = 'No') versus those who left (Attrition = 'Yes').

```python
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

# Determine the number of rows and columns for the subplot grid dynamically
num_numerical_cols = len(numerical_cols)
n_cols = 4 # Number of columns in the subplot grid
n_rows = (num_numerical_cols + n_cols - 1) // n_cols # Calculate number of rows needed

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 4, n_rows * 3), constrained_layout=True)
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    sns.boxplot(x='Attrition', y=col, hue='Attrition', data=df, ax=axes[i], palette={'Yes': 'salmon', 'No': 'lightseagreen'}, legend=False)
    axes[i].set_title(f'{col} by Attrition')
    axes[i].set_xlabel('Attrition')
    axes[i].set_ylabel(col)

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle('Numerical Feature Distributions by Attrition', y=1.02, fontsize=16)
plt.show()
```

### Interpretation of Numerical Features by Attrition

The box plots comparing numerical features against 'Attrition' highlight several distinctions between employees who stayed and those who left:

*   **Age:** Employees who attrite tend to be younger, with a lower median age.
*   **MonthlyIncome:** Attriting employees generally have lower monthly incomes.
*   **DistanceFromHome:** There's a tendency for employees who leave to have a greater distance from home to work.
*   **JobSatisfaction, EnvironmentSatisfaction, RelationshipSatisfaction, WorkLifeBalance:** Employees who leave often report lower satisfaction scores across these categories.
*   **TotalWorkingYears, YearsAtCompany, YearsInCurrentRole, YearsWithCurrManager:** Attriting employees typically have fewer years of total work experience, fewer years at the current company, fewer years in their current role, and fewer years with their current manager.

Features like `DailyRate`, `HourlyRate`, `Education`, `EmployeeNumber`, `PercentSalaryHike`, `PerformanceRating`, and `TrainingTimesLastYear` do not show as stark differences between the two attrition groups.

### Correlation Matrix of Numerical Features

Let's calculate the correlation matrix for all numerical features to understand the linear relationships between them. This can help identify highly correlated features which might be redundant or influential in predicting attrition.

```python
correlation_matrix = df[numerical_cols].corr()
display(correlation_matrix)
```

### Interpretation of the Correlation Matrix

The correlation matrix provides a quantitative measure of the linear relationship between pairs of numerical variables. Key observations include:

*   **Strong Positive Correlations:**
    *   `MonthlyIncome` is highly correlated with `JobLevel` (0.95), and also significantly with `TotalWorkingYears` (0.77).
    *   `TotalWorkingYears` is strongly correlated with `Age` (0.68) and `JobLevel` (0.68).
    *   `YearsAtCompany` shows strong correlations with `YearsInCurrentRole` (0.76), `YearsWithCurrManager` (0.77), and `YearsSinceLastPromotion` (0.62).
*   **Weak/Negligible Correlations:** Many features, such as `DailyRate`, `DistanceFromHome`, `HourlyRate`, `MonthlyRate`, `PercentSalaryHike`, `PerformanceRating`, `TrainingTimesLastYear`, and `WorkLifeBalance`, have very low linear correlations with most other variables.

These correlations suggest natural career progressions where age, experience, job level, and income are interconnected. Also, tenure-related variables within the company are highly intertwined.

### Visualizing the Correlation Matrix with a Heatmap

To better visualize the relationships and identify patterns within the correlation matrix, we'll plot it as a heatmap. This will make it easier to spot highly correlated features at a glance.

```python
plt.figure(figsize=(20, 18))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Numerical Features', fontsize=20)
plt.show()
```

### Interpretation of the Correlation Heatmap

The heatmap visually confirms the relationships identified in the correlation matrix, making them easier to digest:

*   **Bright Red Squares (Strong Positive Correlation):** Highlight pairs like (`MonthlyIncome`, `JobLevel`), (`TotalWorkingYears`, `MonthlyIncome`), and (`YearsAtCompany`, `YearsInCurrentRole`), which signify that these variables tend to increase or decrease together. For example, as `JobLevel` increases, `MonthlyIncome` almost always increases.
*   **Bright Blue Squares (Strong Negative Correlation):** There are no significantly strong negative correlations visible, indicating that no two numerical features are highly inversely related.
*   **Light Colors (Weak/No Correlation):** The majority of the heatmap cells are lighter in color, representing weak or negligible correlations. This includes variables like `DailyRate`, `DistanceFromHome`, `HourlyRate`, and `PerformanceRating` with many other features.

This visualization is crucial for identifying multicollinearity among features, which can be an issue for certain statistical models, and for understanding the inherent structure of the numerical data.
